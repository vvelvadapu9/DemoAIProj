```python
# Vectorization using Bag of Words

from sklearn.feature_extraction.text import CountVectorizer
# Sample corpus

#corpus = ["I love machine learning", "Machine learning is amazing"]
corpus = ["Amazing world requires amazing people - No this and that",
          "I am amazed at how amazing is picked",
          "Funny to see amazing and amazed are different"]
# Initialize CountVectorizer with binary = True
vectorizer = CountVectorizer(binary=True)
#vectorizer = CountVectorizer()
# Fit and transform the corpus (text data)
binary_bow = vectorizer.fit_transform(corpus)
# Convert to array and display results
print("Vocabulary", vectorizer.get_feature_names_out())
print("Binary BoW Representation:\n", binary_bow.toarray())
```

```python
from sklearn.feature_extraction.text import CountVectorizer
# Sample corpus
#corpus = ["I love machine learning",
#          "machine learning is amazing it is giving good learning expereince",
#          "machine are trained with machine learning"]
corpus = ["Amazing world requires amazing people - No this and that",
          "I am amazed at how amazing is picked",
          "Funny to see amazing and amazed are different amazing words"]
# Initialize CountVectorizer with binary = True
vectorizer = CountVectorizer()
# Transform text into BoW representation
bow_matrix = vectorizer.fit_transform(corpus)

#Display results
print("Vocabulary", vectorizer.get_feature_names_out())
print("Binary BoW Representation:\n", bow_matrix.toarray())
```

```python
# TF - IDF (Term Frequency - Inverse document frequency)

from sklearn.feature_extraction.text import TfidfVectorizer
# Sample corpus
corpus = ["Amazing is my love for machine learning",
          "machine are trained with machine learning"]
# Initialize TF-IDF vectorizer
vectorizer = TfidfVectorizer()
# Fit and transform the corpus
tfidf_matrix = vectorizer.fit_transform(corpus)

#display results
print("Vocabulary", vectorizer.get_feature_names_out())
print("TF-IDF Matrix:\n", tfidf_matrix.toarray())
```

# Text Vectorization:
**Bag of Words ** - Limitations
  1) BoW will give high dimensional matrix (document transformational matrix). Even a simple paragraph or few pages corpus ends up with complex high dimensional matrix.
  2) Sparse matrix - resulting in memory wastage
  3) Words are priorized based on frequency
  4) Word meaning has no relevance
  5) Context meaning
  6) Out of vocabulary (OOV) words
  7) Long term dependency

**  TF-IDF** - Limitations
  1) High dimensional like BoW
  2) Sparse matrix
  3) Word meaning has no relevance
  4) Context meaning
  5) Out of Vocabulary (OOV) words
  6) Long term dependency

To overcome above challenges associated with Text vectorization methods we will use Word embedding methods.

Three word embedding methods are 1) Word to Vector 2) Globe Embedding & 3) Fast Text. All these are deep learning models.

1) Word to Vector:
    

```python
txt = '''Once upon a time, in a small village, there lived a kind and hardworking farmer. Every morning,
the farmer would wake up early to tend to his fields and animals. The farmer’s crops were the finest in the land, and his produce was loved by all.
One day, while working in his fields, the farmer discovered a magical stone. The stone had the power to grant one wish to the person who possessed it. The farmer, being wise and
content with his life, wished for happiness for everyone in his village. The magical stone granted his wish, and the entire village was filled with joy and prosperity. From that day forward, the farmer became known as the happiest man in the village. His kindness and generosity were remembered for generations to come.'''

print(txt)

import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

lower = txt.lower()
print(lower)

sw = nltk.corpus.stopwords.words('english')
print(sw)


```

```python
# Word Tokenization & Stopwords removal

from nltk.tokenize import word_tokenize, sent_tokenize
words = word_tokenize(lower)
print(words)
for i in sw:
  if i in words:
    words.remove(i)
word_str = ' '.join(words)
print(word_str)
```

```python
# Sentenece Tokenization
sent = sent_tokenize(word_str)
#print(sent)
sent=''.join(sent)
print(sent)
```

```python
import re
spl_char = re.sub(r'[^\w\s]','', sent)
print(spl_char)
```

```python
words=[]
for word in spl_char:
  words.append(word_tokenize(spl_char))
print(words)
```

2. Building Word Embedding:
    Train a Word2Vec model on the preprocessed text using the Skip-Gram or Continuous Bag of Words(CBOW) approach.
    Use Genesim library to implement Word2Vec model

```python
import gensim
#pip install gensim

from gensim.models import Word2Vec
model = Word2Vec(words, vector_size=100, window=8, min_count=1, sg=1)

# analyze results with different values of vector size, window etc
# vector_size=50
# window=3, 7 and 10 - bigger window size means taking more words into account

```
