```python
import gensim.downloader as api

#Load pre-trained GloVe embeddings (50-dimensional)
glove_vectors = api.load("glove-wiki-gigaword-50")

#Find most similar words
print(glove_vectors.most_similar("king"))
# Ouput: [('prince',0.85), ('queen',0.84), ('monarch', 0.83), ...]

```

```python
from gensim.models import FastText

#Informal and varied corpus
corpus = [
    ["hey", "whatsup", "bro"],
    ["gonna", "grab", "some", "food"],
    ["brb", "need", "to", "charge", "my", "phone"],
    ["lol", "that", "was", "funny"],
    ["omg", "cant", "believe", "this"],
    ["lets", "meet", "at", "the", "cafe"],
    ["u", "coming", "tonight"],
    ["nah", "too", "tired", "maybe", "tomorrow"]
]
```

```python
model = FastText(sentences = corpus,
                 vector_size=100,
                 window=10,
                 min_count=1,
                 epochs=50)

print("Vector for 'grab': ", model.wv['grab'])
```

```python
print("Vector for 'yesterday': ", model.wv['yesterday'])
```

```python
print("Similar to 'whatsup':", model.wv.most_similar('whatsup'))
```

```python
# Try an OOV word with similar morphology
print("Vector for 'whatsuppp':", model.wv['whatsuppp'])
```

```python
from gensim.models.doc2vec import Doc2Vec, TaggedDocument

#Small sample corpus
documents = [
    "The fox jumps over the lazy dog",
    "A quick brown fox is running",
    "Dogs are loyal and friendly",
    "Foxes are clever and fast",
    "I love my dog and my cat"
]

tagged_docs = [TaggedDocument(words=doc.lower().split(),
                              tags=[str(i)]) for i, doc in enumerate(documents)]
```

```python
#Initialize and train the model
model = Doc2Vec(vector_size=50, min_count=1, workers=2, epochs=50)

model.build_vocab(tagged_docs)
model.train(tagged_docs, total_examples=model.corpus_count, epochs=model.epochs)

```

```python
print("Vector for document 0:", model.dv[0])
```

```python
print("Documents similar to doc 0:", model.dv.most_similar('0'))
```

```python
# Infer vector for a new, unseen document
new_doc = "My dog is very fast and clever"

inferred_vector = model.infer_vector(new_doc.lower().split())
print("Inferred vector for new document:", inferred_vector)
```

```python
print("Similar documents to inferred vector:", model.dv.most_similar([inferred_vector]))
```

```python
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

#Sample corpus of sentences
sentences = [
    "The cat sat on the mat",
    "Dogs are loyal animals",
    "I love playing football",
    "Cats and dogs are common pets",
    "Football is a popular sport"
    "Football is my favourite sport"
]

#Tokenize sentences
tokenized = [sentence.lower().split() for sentence in sentences]

model = Word2Vec(sentences=tokenized, vector_size=50, window=5, min_count=1, epochs=100)
```

```python
# sentence embedding function
def sentence_vector(sentence, model):
  words = sentence.lower().split()
  word_vecs = [model.wv[word] for word in words if word in model.wv]
  return np.mean(word_vecs, axis=0) if word_vecs else np.zeros(model.vector_size)
```

```python
#Embed all sentences
sentence_embeddings = [sentence_vector(s, model) for s in sentences]

#User query
query= "who loves playing"
query_vec = sentence_vector(query, model)

#Compute similarity
similarities = cosine_similarity([query_vec], sentence_embeddings)[0]
best_match_index = np.argmax(similarities)


print("Query:", query)
print("Best match:", sentences[best_match_index])
```
