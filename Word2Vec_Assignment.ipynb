```python
docs = [
    "Today I learned about machine learning algorithms",
    "Deep learning uses ANN algorithms",
    "Difference between Deep learning and Machine leaning is one uses ANN while other uses mathematical and statistical algorithms",
    "deep learning and machine learning are so exciting topics",
    "I enjoy learning new things"
]

from gensim.utils import simple_preprocess
tokenized_docs = [simple_preprocess(doc) for doc in docs]
print(tokenized_docs)

from gensim.models import Word2Vec
model = Word2Vec(sentences=tokenized_docs, vector_size=100, window=5, min_count=1, sg=1)

my_learning = model.wv['learning']
print("Vector for 'learning':", my_learning)
print("Vector Sizing:", len(my_learning))
```
