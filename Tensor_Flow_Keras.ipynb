# **Models in Keras/Tensorflow**

**three main types of models in Keras**

- Sequential,

- Functional API,

- Model Scaling


## 1. Sequential Model

**Characteristics:**

- Layers are stacked in a linear order.

- Only supports one input and one output.

- No branching, merging, or shared layers.

**Use Case:**

- Simple feedforward networks like MLPs or basic CNNs.

## 2. Functional API Model

**Characteristics:**

- **Allows non-linear topology:** multiple inputs/outputs, shared layers, skip connections.

- You explicitly define the flow of data between layers.

- More readable than subclassing, more flexible than Sequential.

**Use Case:**

- Complex architectures like ResNet, Siamese networks, multi-task models.


## 3. Model Subclassing

**Characteristics:**

- Full control over forward pass via **call()** method.

- Ideal for dynamic architectures or custom training logic.

- Requires manual layer definition and invocation.

### Use Case:

- Research prototypes, custom loss functions, reinforcement learning.

# **Detailed Explanation of tf.keras.layers.Dense()**

**Syntax:**

- **tf.keras.layers.Dense(**
    - units,
    - activation=None,
    - use_bias=True,
    - kernel_initializer='glorot_uniform',
    - bias_initializer='zeros',
    - kernel_regularizer=None,
    - bias_regularizer=None,
    - activity_regularizer=None,
    - kernel_constraint=None,
    - bias_constraint=None,
    - lora_rank=None,
    - **kwarg)

**Arguments:**

### 1. `units`

- **Type**: `int`
- **Description**: Number of neurons in the layer.
- **Example**: `units=64` creates a layer with 64 output dimensions.
- **Why it matters**: Controls the capacity of the layer.


### 2. `activation`

- **Type**: `str` or callable
- **Description**: Activation function to apply after the linear transformation.
- **Common values**: `'relu'`, `'sigmoid'`, `'tanh'`, `'softmax'`, or `None`
- **Why it matters**: Introduces non-linearity, enabling the network to learn complex patterns.


### 3. `use_bias`

- **Type**: `bool`
- **Default**: `True`
- **Description**: Whether to include a bias term.
- **Why it matters**: Bias allows the model to shift the activation function, improving flexibility.


### 4. `kernel_initializer`

- **Type**: `str` or initializer instance
- **Default**: `'glorot_uniform'` (Xavier initialization)
- **Description**: Method to initialize the weights.
- **Alternatives**: `'he_normal'`, `'random_normal'`, `'zeros'`
- **Why it matters**: Good initialization helps avoid vanishing/exploding gradients.

### 5. `bias_initializer`

- **Type**: `str` or initializer instance
- **Default**: `'zeros'`
- **Description**: Method to initialize the bias vector.
- **Why it matters**: Usually initialized to zero unless specific behavior is desired.


### 6. `kernel_regularizer`

- **Type**: Regularizer instance
- **Description**: Adds penalty (e.g., L1, L2) to the weights during optimization.
- **Example**: `tf.keras.regularizers.l2(0.01)`
- **Why it matters**: Helps prevent overfitting by discouraging large weights.


### 7. `bias_regularizer`

- **Type**: Regularizer instance
- **Description**: Adds penalty to the bias vector.
- **Why it matters**: Rarely used, but can help in very specific regularization schemes.

### 8. `activity_regularizer`

- **Type**: Regularizer instance
- **Description**: Applies penalty to the output of the layer (i.e., activations).
- **Why it matters**: Useful for sparsity constraints or controlling activation magnitude.


### 9. `kernel_constraint`

- **Type**: Constraint instance
- **Description**: Constraint function applied to the weights (e.g., max norm).
- **Example**: `tf.keras.constraints.max_norm(2.0)`
- **Why it matters**: Keeps weights within a certain range, improving stability.

### 10. `bias_constraint`

- **Type**: Constraint instance
- **Description**: Constraint function applied to the bias.
- **Why it matters**: Rarely used, but can enforce bias limits.

### 11. `lora_rank` *(Newer addition)*

- **Type**: `int` or `None`
- **Description**: Used for **Low-Rank Adaptation (LoRA)** — a technique for efficient fine-tuning.
- **Why it matters**: Useful in large models (e.g., transformers) to reduce training cost by adapting only low-rank matrices.

### 12. `**kwargs`
- **Description**: Additional keyword arguments passed to the base Layer class.
- **Why it matters**: Enables customization like naming the layer or setting trainability.

# **4. Compile Model**

- Configure the model for training by specifying **optimizer, loss function, and metrics.**

- Model.compile(
    - optimizer="rmsprop",
    - loss=None,
    - loss_weights=None,
    - metrics=None,
    - weighted_metrics=None,
    - run_eagerly=False,
    - steps_per_execution=1,
    - jit_compile="auto",
    - auto_scale_loss=True,
)



### `optimizer="rmsprop"`

- This sets the optimization algorithm to RMSProp.

- If it passed as a string, so Keras will internally instantiate the default RMSProp optimizer with its standard parameters.

- If you want more control (like learning rate or decay), you could pass an instance like `tf.keras.optimizers.RMSprop(learning_rate=0.001)`.


### `loss=None`

- The `loss` function defines what the model is trying to minimize during training.

- Setting it to `None` means the model won’t know how to compute gradients or update weights.

### `loss_weights=None`

- This is relevant only for multi-output models.

- If your model has multiple outputs (say, predicting both age and gender), you can assign different weights to each loss to control their relative importance.
- Setting it to `None` means all losses are treated equally (default weight = 1).


### `metrics=None`

- Metrics are used to monitor performance during training and evaluation.

- Common examples include `'accuracy'`, `'mae'`, or `'precision'`.

- Setting this to `None` means no metrics will be tracked.


### `weighted_metrics=None`

- This is for metrics that should be computed using sample weights.

- For example, if some samples are more important than others, you can pass weights during training and specify which metrics should respect those weights.

- Leaving it as `None` means no metrics will be weighted.


### `run_eagerly=False`

- This controls whether the model runs in eager mode or graph mode.

- Setting it to `False` enables graph execution, which is faster and more optimized.

- If you were debugging custom layers or training logic, setting this to `True` would make it easier to inspect intermediate values — but at the cost of performance.


### `steps_per_execution=1`

- This defines how many batches to run before updating metrics and logging results.

- A value of `1` means metrics are updated after every batch.

- Increasing this can reduce overhead and improve performance, especially on large datasets or distributed setups.


### `jit_compile="auto"`

- This is a newer feature. It enables XLA (Accelerated Linear Algebra) compilation, which can speed up training by optimizing the computation graph.

- Setting it to `"auto"` lets Keras decide whether to apply JIT compilation based on your hardware and model structure.

- You could also explicitly set it to `True` or `False`.


### `auto_scale_loss=True`

- This is especially useful when training with mixed precision (e.g., float16).

- It automatically scales the loss to prevent numerical instability during backpropagation.

- If you're using GPUs or TPUs with mixed precision enabled, this helps maintain stable gradients and convergence.

# **5. Train Model**

- `model.fit()` trains your model for a fixed number of epochs (iterations over the entire dataset).

- It performs **forward passes, computes loss, backpropagates gradients, and updates weights**.

**Core Syntax:**

- model.fit(
    - x=None,
    - y=None,
    - batch_size=None,
    - epochs=1,
    - verbose=1,
    - callbacks=None,
    - validation_split=0.0,
    - validation_data=None,
    - shuffle=True,
    - class_weight=None,
    - sample_weight=None,
    - initial_epoch=0,
    - steps_per_epoch=None,
    - validation_steps=None,
    - validation_batch_size=None,
    - validation_freq=1,
    - max_queue_size=10,
    - workers=1,
    - use_multiprocessing=False
)

**Key Arguments Explained:**

- **`x` and `y`**: Input data and target labels. Can be NumPy arrays, TensorFlow tensors, or `tf.data.Dataset`.

- **`batch_size`**: Number of samples per gradient update. If not set, defaults to 32.

- **`epochs`**: Number of times the model sees the entire dataset.

- **`verbose`**: Controls output:

  - `0`: Silent
  - `1`: Progress bar
  - `2`: One line per epoch

- **`callbacks`**: List of functions like `EarlyStopping`, `ModelCheckpoint`, or `TensorBoard` that hook into training events.

- **`validation_split`**: Fraction of training data to use for validation (e.g., `0.2` means 20%).

- **`validation_data`**: Explicit validation set as `(x_val, y_val)` or a dataset.

- **`shuffle`**: Whether to shuffle training data before each epoch.

- **`class_weight`**: Dictionary mapping class indices to weights — useful for imbalanced datasets.

- **`sample_weight`**: Array of weights for each training sample.

- **`initial_epoch`**: Start training from this epoch (useful for resuming).

- **`steps_per_epoch`**: Number of batches per epoch (used with generators or datasets).

- **`validation_steps`**: Number of batches to draw from validation data.

- **`validation_batch_size`**: Batch size for validation.

- **`validation_freq`**: How often to evaluate validation data (e.g., every epoch).

- **`max_queue_size`, `workers`, `use_multiprocessing`**: Control data loading performance when using generators.

```python
model.fit()
```

# **6. Evaluating Model**

- `model.evaluate()` runs the model on a test dataset and returns the **loss** and any **metrics** you specified during `compile()`.

- It is used to measure generalization — how well your model performs on data it has not seen before.


**Syntax Overview:**

- model.evaluate(
    - x=None,
    - y=None,
    - batch_size=None,
    - verbose=1,
    - sample_weight=None,
    - steps=None,
    - callbacks=None,
    - return_dict=False,
    - use_multiprocessing=False
)


- **`x`**: Input data — can be a NumPy array, TensorFlow tensor, or `tf.data.Dataset`.

- **`y`**: Target labels corresponding to `x`.

- **`batch_size`**: Number of samples per evaluation batch. Defaults to 32 if not set.

- **`verbose`**: Controls output:

  - `0`: Silent
  - `1`: Progress bar
  - `2`: One line per epoch

- **`sample_weight`**: Optional array of weights for each sample.

- **`steps`**: Number of batches to evaluate (used with generators or datasets).

- **`callbacks`**: List of callback functions to run during evaluation (e.g., logging, monitoring).

- **`return_dict`**: If `True`, returns a dictionary of metric names and values. If `False`, returns a list.

- **`use_multiprocessing`**: If `True`, uses multiple CPU processes for data loading (useful for large datasets).

**When to Use It:**

- After training with `model.fit()`

- To benchmark performance on test data

- To compare models or training strategies
