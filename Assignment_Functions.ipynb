```python
#3 function find_max(a, b) that returns the maximum of two numbers a and b

def find_max(a, b):
    if a > b:
        return a
    else:
        return b

print(find_max(16, 27))

```

```python
#4 function count_vowels(s) that returns the number of vowels (a, e, i, o, u) in a string s.

def count_vowels(s):
  vowels = "aeiou"
  count = 0
  s = s.lower()
  for char in s:
        if char in vowels:
            count = count + 1
  return count

print(count_vowels("I know all vowels a E i O u"))
```

```python
#5 function concat_strings(s1, s2) that returns the concatenation of two strings s1 and s2.

def concat_strings(s1, s2):
    concat = s1 + s2
    return concat

print(concat_strings("I am David Johnson", " from Dallas, Texas."))
```

```python
#6 function remove_duplicates(lst) that takes a list lst and returns a new list with duplicate elements removed.

def remove_duplicates(lst):
    new_list = []
    for item in lst:
        if item not in new_list:
            new_list.append(item)
    return new_list

print(remove_duplicates([23, 23, 54, 45, 9, 9, 81, 90, 38, 34, 45]))

```

```python
#8 Write a function sum_even_numbers(lst) that returns the sum of all even numbers in a list lst.

def sum_even_numbers(lst):
    even_list = []
    sum = 0
    for num in lst:
        if num % 2 == 0:
            sum = sum + num
            even_list.append(num)
    #print(even_list)
    return sum

print(sum_even_numbers([23, 2, 16, 45, 9, 9, 81, 90, 38, 34, 45]))
```

```python
# function char_frequency(s) returning dictionary with keys are characters in the string and their values as number of times characters appear.

def char_frequency(s):
    char_count = {}
    for char in s:
        if char in char_count:
            char_count[char] = char_count[char] + 1
        else:
            char_count[char] = 1
    return char_count

print(char_frequency("Its Easy to find easy answers for hard problems"))
```
