```python
#Q1 - slice and indexes
x=["Earth", "revolves", "around" ,"the", "sun"]
print(x)
print(x[1]) #positive index for elements start from 0
print(x[-1]) #negative index starts from -1 for the last element

#string from middle two characters
y=str(input("Enter your string:"))
#print(y)
if(len(y))%2 == 0:
  print(y[len(y)//2-1:len(y)//2+1])
else:
  print(y[len(y)//2])
```

```python
#Q2 string's first, middle and last
z=str(input("Enter your string:"))
a=z[0]

if(len(z))%2==0:
  b=z[len(z)//2-1:len(z)//2+1]
else:
  b=z[len(z)//2]

c=z[-1]

print(a+b+c)
```

```python
#Q3 Five different string methods
#Upper, Replace, Split, Ascii letters, digits

#Upper/Lower/Title and Capitalize
m = "ravidra kumar"
print(m.upper()) #Upper method converts string to upper case

n = "RaVinDra KumAR"
print(n.lower()) #Lower method converts string to lower case

t = "raVinDra kuMar adma"
print(t.title()) #Title method starts with Upper case for ever word

c = "ravidra KumAR Adma"
print(c.capitalize()) #Capitalize method starts only first character in Upper case

#Replace is used to replace part of a string with another string

r = "Mary had a little lamb"

#Need to replace Mary with John & Lamb with Pony in above string
r=r.replace("Mary", "John")
r=r.replace("lamb", "Pony")
print(r)

#Splits a string into a list based on a separator
s = "Innomatics Data Labs"
print(s.split())

#To provide all ASCII letters and digits
import string
print(string.ascii_letters)
print(string.ascii_lowercase)
print(string.ascii_uppercase)
print(string.digits)
```

```python
#Q4 palindrome
p=str(input("Enter your string:"))
#print(p)
prev=p[::-1]
#print(prev)
if p==prev:
  is_palindrome = True
  print("Your string is a palindrome")
else:
  is_palindrome = False
  print("Your string is not a palindrome")

#Capitalize
x = str(input("Enter your string: "))
print(x.capitalize())
```
