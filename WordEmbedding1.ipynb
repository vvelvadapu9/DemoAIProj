```python
#!pip install gensim

from gensim.models import Word2Vec
text=["the", "cat", "sits", "on", "the", "mat"], ["the", "dog", "plays", "in", "the", "garden"],["cats", "and", "dogs", "are", "friends"]
#print(text)

model = Word2Vec(text,vector_size=100, window=5, min_count=1, sg=1)
vector = model.wv['cat']
print("Vector for 'Cat':" ,vector)
print(len(vector))

```

```python
vector = model.wv['the']
print("Vector for 'the':" ,vector)
print(len(vector))
```

```python
#Find the most similar words to 'cat'
similar_words = model.wv.most_similar('cat', topn=5)
print("Most similar words to 'cat':", similar_words)
print(len(similar_words))
```

```python
text = """
Barack Obama was born on August 4, 1961, in Honolulu, Hawaii.He served as the 44th President of the United States from 2009 to 2017.
Apple Inc., headquartered in Cupertino, California, is a leading technology company.
"""
print(text)
```

```python
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize
words = word_tokenize(text)
print(words)
```

```python
# Parts of Speech (POS) tagging
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk import pos_tag
pos_tags = pos_tag(words)
print(pos_tags)
```

```python
nltk.download('maxent_ne_chunker_tab')
from nltk import ne_chunk
```

```python
ner=ne_chunk(pos_tags)
print(ner)
```
