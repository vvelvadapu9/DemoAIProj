```python
from gensim.models import FastText

#Given Corpus
corpus = """Generative AI (GenAI) represents one of the most transformative areas in artificial intelligence, focusing on the ability of machines to create new content such as text, images, audio, and video.
 Companies across industries are rapidly adopting GenAI to automate creative processes, enhance productivity, and personalize user experiences.
Modern GenAI systems are built on large language models (LLMs) like GPT, Llama, and Gemini.
 These models are trained on massive datasets using self-supervised learning and fine-tuned for specific applications such as chatbots, code generation, and content creation.
Enterprises use tools like LangChain, Hugging Face Transformers, and Prompt Engineering techniques to integrate GenAI into their workflows.
 For example, marketing teams use GenAI for ad copy generation, developers use it for code assistance, and educators use it to create personalized learning materials.
Training data quality, tokenization, embeddings, and model architecture play a critical role in determining the quality of generated outputs.
 To improve accuracy, researchers employ vector databases, retrieval-augmented generation (RAG), and fine-tuning pipelines to make AI systems more context-aware.
GenAI has also influenced fields like design, healthcare, and education by enabling faster idea generation and adaptive feedback systems.
 Ethical AI and bias mitigation remain major focus areas to ensure responsible AI development.
Overall, Generative AI continues to redefine creativity and innovation by bridging the gap between human imagination and machine intelligence."""


#Processing this text
sentences = [sentence.lower().split() for sentence in corpus.split('.') if sentence.strip()]

#Train the model

fasttext_model = FastText(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=1,              # chnage to 0 for CBOW
    epochs=50
)

print("Vector for AI", fasttext_model.wv['ai'])

```

```python
#Analyze similarity between key terms:
#"AI" vs "machine"
print("Similarity between AI and Machine:", fasttext_model.wv.similarity('ai', 'machine'))

#"creativity" vs "innovation"
print("Similarity between Creativity and Innovation:", fasttext_model.wv.similarity('creativity', 'innovation'))

#"learning" vs "training"
print("Similarity between Learning and Training:", fasttext_model.wv.similarity('learning', 'training'))

#"models" vs "embeddings"
print("Similarity between Models and Embeddings:", fasttext_model.wv.similarity('models', 'embeddings'))
```

```python
#Prepare 4–5 short documents (rows) related to the GenAI context.
rows = [
    "Generative AI models can create text images and code",
    "Large language models are trained on massive datasets",
    "FastText and Word2Vec learn word embeddings",
    "Neural networks generate human like responses",
    "Artificial intelligence improves content generation"
]

sentences = [doc.lower().split() for doc in rows]
#print(sentences)


#Convert to vector
model = FastText(sentences=rows,
                 vector_size=100,
                 window=10,
                 min_count=1,
                 sg=1,
                 epochs=50)
print("Vector for 'models'", model.wv['models'])
```

```python

#Compare vector similarities between selected words

print("Similarity between 'human' and 'responses':", model.wv.similarity('human', 'responses'))

print("Similarity between 'images' and 'data':", model.wv.similarity('images', 'data'))

print("Similarity between 'neural' and 'networks':", model.wv.similarity('neural', 'networks'))
```
