```python
#name that shape
s=int(input("Enter the number of sides of shape: "))
if s<3 or s>10:
  print("Input is not correct. Enter side between 3 and 10")
elif s == 3:
  print("This is a triangle")
elif s == 4:
  print("This is a quadrilateral")
elif s == 5:
  print("This is a pentagon")
elif s == 6:
  print("This is a hexagon")
elif s == 7:
  print("This is a heptagon")
elif s == 8:
  print("This is a octogon")
elif s == 9:
  print("This is a nonagon")
elif s == 10:
  print("This is a decagon")
```

```python
# Program to display the shape based on number of sides
# Dictionary mapping number of sides to shape names

shapes = {
    3: "Triangle",
    4: "Quadrilateral",
    5: "Pentagon",
    6: "Hexagon",
    7: "Heptagon",
    8: "Octagon",
    9: "Nonagon",
    10: "Decagon"
}

# Input from user
sides = int(input("Enter number of sides (3–10): "))

# Check and display the shape
if 3 <= sides <= 10:
    print("The shape is:", shapes[sides])
else:
    print("Invalid input! Please enter a number between 3 and 10.")
```
