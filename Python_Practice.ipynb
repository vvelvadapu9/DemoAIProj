```python
#Tuple Function
#Tuple is list of elements similar to List however data within Tuple cannot be changed/modified. Tuple is immutable
x=(10,20,30,540,50)
t1=("hyd", "chn", "blr", "del")
t2=("vj",123,"yz",999)
print(len(x))
print(type(x))

#function with Tuple - count, len, index,
print(x.count(10))
print(x.index(10))

print(t1.count("del"))
print(t2.count(123))
```

```python
#Dictionary - It stores key value pairs;
#Dictionary is called a mapping data type

#Creating a dictionary
dic={"English":92, "Math":99, "Science":95, "History":94}
print(dic)
print(type(dic))

rollc={"R1":"David", "R2":123, "R3":"Maggie"}
print(rollc)

#Dictionary with list as value
runs={"virat":[35,68,120,5,12],
      "MSD":[34,80,12,34,44],
      "ABD":[]
    }
print(len(runs))

#functions clear, del,
dic.clear()
print(dic)
dic={"English":92, "Math":99, "Science":95, "History":94}
print(dic)
#del dic
print(dic)
#get allows retreiving data
print(dic.get("English"))
print(dic.get("History"))

# dict(): if data is given as list of tuples, dict() function converts into dictionary
data = [(1,"a"), (2,"b"), (3,"c"), (4,"d") ]
d=dict(data)
print(d)

#keys(): It returns all the keys from distionary
print(d)
print(d.keys())

# values(): it returns all values associated with dictionary
print(d)
print(d.values())

#sets - allows to remove duplicates

s={10,20,30,5,50,12,32}
print(s)

s2={10,20,10,20,10,20}
print(s2)

#Python Operators
n1=24
n2=20
n3=2
print(n1-n2)
print(n1+n2)
print(n1*n2)
print(n1/n2) #returns quotient as float
print(n1//n2) #returns quotient as integer
print(n1%n2) #returns remainder
print(n2**n3) #returns exponential value


```
