```python
#12 palindrome program

s=str(input("Enter any string(oneword): "))
s = s.lower()
i=0
j=len(s)-1
ispalindrome = True
while i<j:
  if s[i]!=s[j]:
    ispalindrome = False
    break
  i=i+1
  j=j-1

if ispalindrome:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```

```python
#10 Validate input 3 sides to determine triangle is possible

def istriangle(a,b,c):
  istri = 0
  if (a+b)>c and (b+c)>a and (c+a)>c:
    istri = True
    print("Sides ", a, b, c, "form a triangle")
  else:
    istri = False
    print("Sides ", a, b, c, "cannot form a triangle")
  return(istri)

print(istriangle(3,4,5))

print(istriangle(6,8,18))

# without functions

a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))

if (a+b)>c and (b+c)>a and (c+a)>c:
  print("Sides ", a, b, c, "form a triangle")
else:
  print("Sides ", a, b, c, "cannot form a triangle")

```

```python
#8 Calculate bonus based on years of service

salary = float(input("Enter your salary: "))
years = float(input("Enter your years of service: "))

if years > 5:
    bonus = salary * 5/100
    print("Congratulations! Your bonus is:", bonus)
else:
    print("No bonus. Years of service must be more than 5.")
```

```python
#7 Check word is plural or not based on last character

plural_word = str(input("Enter your word: "))
if plural_word.endswith("s") or plural_word.endswith("S"):
  print("Your word is plural")
else:
  print("Your word is not plural")
```

```python
#6 Add suffix of ing or ly based on lenght of string

word = str(input("Enter your word: "))
if len(word) < 3:
  print(word)
elif word.endswith("ing"):
  print(word + "ly")
else:
  print(word + "ing")
```
