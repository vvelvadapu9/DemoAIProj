```python
n=int(input("Enter the beginning number"))
sum=0
count=0
while n!=0:
  sum=sum+n
  count=count+1
  n=int(input("Enter the next number"))

print("Sum of all numbers provided = ", sum)
avg = sum/count
print("Average of all numbers provided =", avg)
```

```python
original = 4.95
discount = 60/100

print ("Original  \tNew Price \tDiscount")
while original <=25:
  newprice = original * discount
  newprice = round(newprice,2)
  discprice= original - newprice
  discprice = round(discprice,2)
  print("$",original, "\t$",newprice, "\t$", discprice)
  original = original+5
```

```python
#Zoo Admission Price
age=input("Enter the age of first person")
price = 0
while age!= "":
  age=int(age)
  if age<=2:
    price= price+0
  elif age>2 and age<=12:
    price = price +14
  elif age>=65:
    price = price +18
  else:
    price=price+23
  age=input("Enter age of next person")
print("The total price for Admission is $", price)
```

```python
#Approximate Pie

pi=3
k=2
i=2
while i<31:
  pi = pi + (((-1)**k)*4)/ ((i)*(i+1)*(i+2))
  i=i+2
  k=k+1
print(f"value of pi after 15 approximations is {pi}")
```

```python
#find square root of a number
n=int(input("Enter any positive integer"))
guess=n/2
while ((guess*guess)-n)>1e-12:
  guess=(guess+(n/guess))/2
print(f"square root of {n} is {guess}")
```

```python
# find greatest common divisor

m=int(input("Enter first number"))
n=int(input("Enter second number"))
if m>n:
  d=n
else:
  d=m

while m%d !=0 or n%d !=0 :
  d=d-1
print(f"Greatest Common Divisor is {d}")
```

```python
#palindrome program

s=str(input("Enter any string(oneword)"))
i=0
j=len(s)-1
while True:
  if s[i]!=s[j]:
    print("Its not a palindrome")
    break
  else:
    if j==0:
      print("Its a palindrome")
    i=i+1
    j=j-1
```
