```python


def fib(n):    # write Fibonacci series less than n
    """Print a Fibonacci series less than n."""
    print ("Below is Fibonacci series for numbers upto -", n)

    a, b = 0, 1
    while a < n:
        print(a, end=' ')
        a, b = b, a+b
    print()

fib(500)
```

```python
#Working with lists

x = [10,20,30,10,40,50]
y = ["hyd","blr","chn","del"]
z = ["Rich", 10, 123, 345, "xyz"]
w = [100,200,300,400,500]
print(x)
print(type(x))
print(len(x))

print(x+y)
print(x*2)
print(2*y)

#slicing --> Lets us access one element from a list
print(w[2]) #positive index for elements start from 0
print(z[-3]) #negative index starts from -1 for the last element
print(y[0:4])
print(y[-3:-1])

#accessing alternating elements
print(x[0:5:2]) #starts from 0 upto 5th element giving alternates
print(x[1:len(x):2]) #starts from 1 upto lenght of list providing alternates

#list operations - append, insert, extend, remove, pop, reverse, sort

print(x.index(10))
x.append("60") #append allows to add new elements at the end, one at a time
print(x)
x.insert(4,20) #insert allows to add new elements at specific index
print(x)
x.extend(w) #extend concatenates w to x list (similar to x+w)
print(x)
x.remove(40) #removes first occurence of element passed
print(x)
x.pop(4) #pop removes specific index value element from the list
print(x)


print(w)
w.reverse()#reverse allows to reverse the order of elements in the list
print(w)
#sort allows to sort lists with same data types
w.sort()
print(w)
#w.sorted()
#print(w)

y.sort(reverse=True)
print(y)
```
