```python
#Handling Basic Exceptions:
try:
    num = int(input("Enter a number: "))
    result = 10 / num
    print("Result: ", result)
except ZeroDivisionError:
    print("Error: Division by zero is not allowed")
except ValueError:
    print("Error: Invalid input. Please enter a valid number.")
except Exception as e:
    print("An error occurred:", e)
```

```python
#Handling Multiple Exceptions
try:
    numbers = [1,2,3]
    index = int(input("Enter an index: "))
    result = numbers[index]
    print("Number: ", result)
except IndexError:
    print("Error: Index out of range")
except ValueError:
    print("Error: Invalid input. Please enter a valid integer.")
```

```python
# try - except - finally
try:
    num = int(input("Enter a number: "))
    print("You entered:", num)
except ValueError:
    print("Error: Invalid input!")
else:
    print("No exceptions occurred")
finally:
    print("This always executes.")
```

```python
#Let us create a person class
class person():
  # Let us add objects using _inti_construtor.
  def __init__(self, name, age, height, nation):
    self.name = name
    self.age = age
    self.height = height
    self.nation = nation
  # Let us create a method to print name of person
  def pname(self):
    print(self.name)
  def page(self):
    print(self.age)
  def pheight(self):
    print(self.height)
  def pnation(self):
    print(self.nation)
  def pdetails(self):
    print("Name:", self.name)
    print("Age:", self.age)
    print("Height:", self.height)
    print("Nationality:", self.nation)

# Create instance of person class
p1 = person("tom",44,180,"USA")
p2 = person("david", 33, 170, "India")

print(p1.pdetails())
print(p2.pdetails())
```

```python
class calc():
  def __init__(self, a, b):
    self.a = a
    self.b = b
  def addition(self):
    print("sum: ", self.a+self.b)
  def difference(self):
    print("difference: ", self.a-self.b)
  def product(self):
    print("product: ", self.a*self.b)
  def division(self):
    print("division: ", self.a/self.b)

p = calc(10,20)

print(p.addition())
print(p.difference())
print(p.product())
print(p.division())
```

```python
# distance between two points
import math
class distance():
  def __init__(self, x1, y1, x2, y2):
    self.x1 = x1
    self.y1 = y1
    self.x2 = x2
    self.y2 = y2
  def point_distance(self):
    distance = math.sqrt((self.x2 - self.x1)**2 + (self.y2 - self.y1)**2)
    print(f"The distance between the points ({self.x1}, {self.y1 }) & ({self.x2}, {self.y2}) is:",  distance)


p=distance(3,4,6,8)
print(p.point_distance())

```

```python
# let us create first class
class one():
  def __init__(self, num1, num2):
    self.num1 = num1
    self.num2 = num2
  def numbers(self):
    print(self.num1, self.num2)

#let us create second class
class two():
  def __init__(self, name1, name2):
    self.name1 = name1
    self.name2 = name2
  def names(self):
    print(self.name1, self.name2)

cls1 = one(34,45)
cls2 = two("tom", "john")

print(cls1.numbers())
print(cls2.names())
```
